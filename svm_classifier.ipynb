# Train the model:

```python
from sklearn.svm import SVC
```

```python
model = SVC(kernel="linear", C=1000)
```

```python
dataset.target
```

```python
y = dataset.target[:100]
```

```python
print(y.shape)
print(f"\n{len(y)}")
```

# Starting with X_slw

```python
X_slw = X_slw[:100]
X_slw.shape
```

```python
plt.scatter(X_slw[:,0], X_slw[:,1], c=y)
plt.show()
```

```python
model.fit(X_slw, y)
```

```python
score = model.score(X_slw, y) * 100
print(f"Model's score is: {score}%")
```

```python
model.decision_function(X_slw)
```

```python
len(model.decision_function(X_slw))
```

```python
dist = model.decision_function(X_slw)
max_dist = max(dist)

max_negative_dist = min(dist)

#To print the index of the values; use np.where() it returns the tuple of arrays.
result1 = np.where(dist == max_dist)
idx1 = result1[0][0]
value = X_slw[idx1]
print(f"Index is: {idx1} and the max-positive-side value is: {value} which is {max_dist}")

result2 = np.where(dist == max_negative_dist)
idx2 = result2[0][0]
value2 = X_slw[idx2]
print(f"Index is: {idx2} and the max-negative-side value is: {value2} which is {max_negative_dist}")
```

```python
model.coef_[0][1] # Value of W, which is the margin
```

```python
margin = 1 / model.coef_[0][1]
n_margin = np.abs(margin)
print(n_margin)
```
