```python
import sklearn
```

```python
from sklearn.datasets import fetch_openml 
```

```python
dataset = fetch_openml('mnist_784', as_frame = False)
```

```python
X = dataset['data']
y = dataset['target']
```

```python
y = y.astype(int)
```

```python
#converting multiclass to binary class
```

```python
y_bin = (y == 5)
```

```python
y_bin
```

```python
y_bin[0]
```

```python
y_bin[:10]
```

```python
#Split the dataset into Train and Test

X_train = X[ :60000]
X_test = X[60000: ]

y_train = y_bin[ : 60000]
y_test = y_bin[60000: ]

X_train.shape
```

```python
X_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
from sklearn.linear_model import SGDClassifier
```

```python
model = SGDClassifier(random_state = 42)
```

```python
#Time to Start Training
# using fit() to train the model 

model.fit(X_train, y_train)
```

```python
#Time to Predict 

model.predict([X[0]])
```

```python
model.predict([X[1]])
```

```python
model.predict([X[2]])
```

```python
model.predict([X_test[0]])
```

```python
from sklearn.metrics import confusion_matrix
```

```python
confusion_matrix(y_train, model.predict(X_train))
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score
```

```python
precision_score(y_train, model.predict(X_train))
```

```python
recall_score(y_train, model.predict(X_train))
```

```python
f1_score(y_train, model.predict(X_train))
```
